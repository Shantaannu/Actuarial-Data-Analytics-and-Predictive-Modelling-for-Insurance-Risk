```python
# ============================================
# SECTION 3: DATA CLEANING & PREPARATION
# ============================================

# 1. Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python

# ============================================
# 2. Load the datasets
# ============================================

claims = pd.read_csv("../data/raw/ncid_private_motor_ultimate_claims.csv")
premium = pd.read_csv("../data/raw/ncid_private_motor_ultimate_prem_claims.csv")

```

```python
# ============================================
# 3. Create working copies
# ============================================

claims_clean = claims.copy()
premium_clean = premium.copy()
```

```python
# ============================================
# 4. Check dataset size
# ============================================

print("Claims dataset shape:", claims_clean.shape)
print("Premium dataset shape:", premium_clean.shape)
```

```python

# ============================================
# 5. View first 5 rows
# ============================================

print("\n--- Claims Data ---")
display(claims_clean.head())

print("\n--- Premium Data ---")
display(premium_clean.head())
```

```python

# ============================================
# 6. Check column names
# ============================================

print("\nClaims columns:")
print(claims_clean.columns.tolist())

print("\nPremium columns:")
print(premium_clean.columns.tolist())
```

```python
# ============================================
# 7. Check data types
# ============================================

print("\n--- Claims Data Types ---")
print(claims_clean.dtypes)

print("\n--- Premium Data Types ---")
print(premium_clean.dtypes)
```

```python
# ============================================
# 8. Check missing values
# ============================================

print("\n--- Missing Values: Claims ---")
print(claims_clean.isnull().sum())

print("\n--- Missing Values: Premium ---")
print(premium_clean.isnull().sum())
```

```python

# ============================================
# 9. Check duplicate rows
# ============================================

print("\nDuplicate rows in Claims:",
      claims_clean.duplicated().sum())

print("Duplicate rows in Premium:",
      premium_clean.duplicated().sum())
```

```python

# ============================================
# 10. Check unique Measures
# ============================================

print("\n--- Claims Measures ---")
print(claims_clean["Measure"].unique())

print("\n--- Premium Measures ---")
print(premium_clean["Measure"].unique())
```

```python


# ============================================
# 11. Count each Measure
# ============================================

print("\n--- Claims Measure Counts ---")
print(claims_clean["Measure"].value_counts())

print("\n--- Premium Measure Counts ---")
print(premium_clean["Measure"].value_counts())


# ============================================
# 12. Check Claim Types
# ============================================

print("\n--- Claim Types ---")
print(claims_clean["ClaimType"].unique())

print("\nClaim Type Counts:")
print(claims_clean["ClaimType"].value_counts())
```

```python



# ============================================
# 13. Check Cover Types
# ============================================

print("\n--- Cover Types ---")
print(premium_clean["CoverType"].unique())

print("\nCover Type Counts:")
print(premium_clean["CoverType"].value_counts())



# ============================================
# 14. Check year range
# ============================================

print("\nClaims year range:")
print(claims_clean["Year"].min(), "to", claims_clean["Year"].max())

print("\nPremium year range:")
print(premium_clean["Year"].min(), "to", premium_clean["Year"].max())


# ============================================
# 15. Check AccidentQuarter
# ============================================

print("\n--- Accident Quarter Examples ---")
print(claims_clean["AccidentQuarter"].unique()[:10])

# ============================================
# 16. Create Quarter variable
# ============================================

claims_clean["Quarter"] = (
    claims_clean["AccidentQuarter"]
    .astype(str)
    .str[-2:]
    .astype(int)
)

```

```python

# ============================================
# 17. Create readable Quarter Label
# ============================================

claims_clean["Quarter_Label"] = (
    claims_clean["Year"].astype(str)
    + " Q"
    + claims_clean["Quarter"].astype(str)
)


# Check result
print("\n--- Quarter Labels ---")
display(
    claims_clean[
        ["Year", "AccidentQuarter", "Quarter", "Quarter_Label"]
    ].head(10)
)


# ============================================
# 18. Separate Ultimate Claim Costs
# ============================================

claim_costs = claims_clean[
    claims_clean["Measure"] == "Ultimate Costs"
].copy()


# ============================================
# 19. Separate Ultimate Claim Numbers
# ============================================

claim_numbers = claims_clean[
    claims_clean["Measure"] == "Ultimate Numbers (incl. nils)"
].copy()
```

```python

# ============================================
# 20. Separate Gross Earned Premium
# ============================================

earned_premium = premium_clean[
    premium_clean["Measure"] == "Gross Earned Premium"
].copy()


# ============================================
# 21. Separate Earned Policy Count
# ============================================

policy_count = premium_clean[
    premium_clean["Measure"] == "Earned Policy Count"
].copy()


# ============================================
# 22. Check resulting datasets
# ============================================

print("Ultimate Claim Costs:", claim_costs.shape)
print("Ultimate Claim Numbers:", claim_numbers.shape)
print("Gross Earned Premium:", earned_premium.shape)
print("Earned Policy Count:", policy_count.shape)
```

```python

# ============================================
# 23. Check numerical statistics
# ============================================

print("\n--- Claim Cost Statistics ---")
display(claim_costs["Value"].describe())

print("\n--- Claim Number Statistics ---")
display(claim_numbers["Value"].describe())

print("\n--- Premium Statistics ---")
display(earned_premium["Value"].describe())

print("\n--- Policy Count Statistics ---")
display(policy_count["Value"].describe())
```

```python
# ============================================
# 24. Check for negative values
# ============================================

print("\nNegative Claim Costs:",
      (claim_costs["Value"] < 0).sum())

print("Negative Claim Numbers:",
      (claim_numbers["Value"] < 0).sum())

print("Negative Premium:",
      (earned_premium["Value"] < 0).sum())

print("Negative Policy Counts:",
      (policy_count["Value"] < 0).sum())
```

```python

# ============================================
# 25. Final preview
# ============================================

print("\n--- Clean Claims Dataset ---")
display(claims_clean.head())

print("\n--- Clean Premium Dataset ---")
display(premium_clean.head())
```
