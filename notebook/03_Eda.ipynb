```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
claims = pd.read_csv("../data/raw/ncid_private_motor_ultimate_claims.csv")

premium = pd.read_csv("../data/raw/ncid_private_motor_ultimate_prem_claims.csv")
```

```python
claim_costs = claims[
    claims["Measure"] == "Ultimate Costs"
].copy()
```

```python
claim_numbers = claims[
    claims["Measure"] == "Ultimate Numbers (incl. nils)"
].copy()
```

```python
earned_premium = premium[
    premium["Measure"] == "Gross Earned Premium"
].copy()

policy_count = premium[
    premium["Measure"] == "Earned Policy Count"
].copy()
```

```python
annual_claims = (
    claim_costs
    .groupby("Year")["Value"]
    .sum()
    .reset_index()
)

annual_claims.columns = [
    "Year",
    "Ultimate_Claims"
]

annual_claims
```

```python
#Annual Ultimate Claim Costs
annual_claims = (
    claim_costs
    .groupby("Year")["Value"]
    .sum()
    .reset_index()
)

annual_claims.columns = ["Year", "Ultimate_Claims"]

annual_claims
```

```python
#Plot Ultimate Claims Over Time
plt.figure(figsize=(12, 6))

plt.plot(
    annual_claims["Year"],
    annual_claims["Ultimate_Claims"],
    marker="o"
)

plt.title("Annual Ultimate Motor Insurance Claims")
plt.xlabel("Year")
plt.ylabel("Ultimate Claims")
plt.grid(True)

plt.show()
```

```python
#4.4 Annual Claim Numbers
annual_claim_numbers = (
    claim_numbers
    .groupby("Year")["Value"]
    .sum()
    .reset_index()
)

annual_claim_numbers.columns = ["Year", "Ultimate_Claim_Numbers"]

annual_claim_numbers
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    annual_claim_numbers["Year"],
    annual_claim_numbers["Ultimate_Claim_Numbers"],
    marker="o"
)

plt.title("Annual Ultimate Number of Motor Insurance Claims")
plt.xlabel("Year")
plt.ylabel("Number of Claims")
plt.grid(True)

plt.show()
```

```python
#Claim Type Analysis
claims_by_type = (
    claim_costs
    .groupby("ClaimType")["Value"]
    .sum()
    .sort_values(ascending=False)
)

claims_by_type
```

```python
plt.figure(figsize=(12, 6))

claims_by_type.plot(kind="bar")

plt.title("Ultimate Claims by Claim Type")
plt.xlabel("Claim Type")
plt.ylabel("Ultimate Claim Cost")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
#Claim Numbers by Claim Type
claim_numbers_by_type = (
    claim_numbers
    .groupby("ClaimType")["Value"]
    .sum()
    .sort_values(ascending=False)
)

claim_numbers_by_type
```

```python
plt.figure(figsize=(12, 6))

claim_numbers_by_type.plot(kind="bar")

plt.title("Ultimate Number of Claims by Claim Type")
plt.xlabel("Claim Type")
plt.ylabel("Number of Claims")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
#4.7 Premium Analysis
annual_premium = (
    earned_premium
    .groupby("Year")["Value"]
    .sum()
    .reset_index()
)

annual_premium.columns = ["Year", "Gross_Earned_Premium"]

annual_premium
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    annual_premium["Year"],
    annual_premium["Gross_Earned_Premium"],
    marker="o"
)

plt.title("Annual Gross Earned Premium")
plt.xlabel("Year")
plt.ylabel("Gross Earned Premium")
plt.grid(True)

plt.show()
```

```python
#4.8 Policy Count Analysis
annual_policies = (
    policy_count
    .groupby("Year")["Value"]
    .sum()
    .reset_index()
)

annual_policies.columns = ["Year", "Earned_Policy_Count"]

annual_policies
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    annual_policies["Year"],
    annual_policies["Earned_Policy_Count"],
    marker="o"
)

plt.title("Annual Earned Policy Count")
plt.xlabel("Year")
plt.ylabel("Number of Policies")
plt.grid(True)

plt.show()
```

```python
#4.9 Combine Claims and Premium
annual_data = annual_claims.merge(
    annual_claim_numbers,
    on="Year",
    how="outer"
)

annual_data = annual_data.merge(
    annual_premium,
    on="Year",
    how="outer"
)

annual_data = annual_data.merge(
    annual_policies,
    on="Year",
    how="outer"
)

annual_data
```

```python
# Claim Frequency

annual_data["Claim_Frequency"] = (
    annual_data["Ultimate_Claim_Numbers"]
    / annual_data["Earned_Policy_Count"]
)

annual_data[["Year", "Claim_Frequency"]]

annual_data["Claim_Frequency_%"] = (
    annual_data["Claim_Frequency"] * 100
)

annual_data[["Year", "Claim_Frequency_%"]]
```

```python
# Average Claim Cost

annual_data["Average_Claim_Cost"] = (
    annual_data["Ultimate_Claims"]
    / annual_data["Ultimate_Claim_Numbers"]
)

annual_data[
    ["Year", "Average_Claim_Cost"]
]
```

```python
# Loss Ratio

annual_data["Loss_Ratio"] = (
    annual_data["Ultimate_Claims"]
    / annual_data["Gross_Earned_Premium"]
) * 100

annual_data[
    ["Year", "Loss_Ratio"]
]
```

```python
actuarial_analysis = annual_data[
    [
        "Year",
        "Ultimate_Claims",
        "Ultimate_Claim_Numbers",
        "Gross_Earned_Premium",
        "Earned_Policy_Count",
        "Claim_Frequency_%",
        "Average_Claim_Cost",
        "Loss_Ratio"
    ]
].copy()

actuarial_analysis
```

```python
#Find the highest Loss Ratio years
actuarial_analysis.sort_values(
    "Loss_Ratio",
    ascending=False
).head(5)
```

```python
#Find the highest average claim cost
actuarial_analysis.sort_values(
    "Average_Claim_Cost",
    ascending=False
).head(5)
```

```python
#Find the highest claim frequency
actuarial_analysis.sort_values(
    "Claim_Frequency_%",
    ascending=False
).head(5)
```

```python
annual_claims = (
    claim_costs
    .groupby("Year")["Value"]
    .sum()
    .reset_index()
)

annual_claims.columns = [
    "Year",
    "Ultimate_Claims"
]

annual_claims
```

```python
#Calculate Loss Ratio
claims_premium["Loss_Ratio"] = (
    claims_premium["Ultimate_Claims"]
    / claims_premium["Gross_Earned_Premium"]
) * 100

claims_premium
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    claims_premium["Year"],
    claims_premium["Loss_Ratio"],
    marker="o"
)

plt.title("Annual Loss Ratio")
plt.xlabel("Year")
plt.ylabel("Loss Ratio (%)")

plt.grid(True)
plt.show()
```

```python
highest_loss_ratio = (
    claims_premium
    .sort_values("Loss_Ratio", ascending=False)
    .head(5)
)

highest_loss_ratio
```

```python
lowest_loss_ratio = (
    claims_premium
    .sort_values("Loss_Ratio", ascending=True)
    .head(5)
)

lowest_loss_ratio
```

```python
def classify_loss_ratio(loss_ratio):
    if loss_ratio < 60:
        return "Low"
    elif loss_ratio < 80:
        return "Medium"
    else:
        return "High"

claims_premium["Risk_Category"] = (
    claims_premium["Loss_Ratio"]
    .apply(classify_loss_ratio)
)

claims_premium[
    ["Year", "Loss_Ratio", "Risk_Category"]
]
```

```python
risk_summary = (
    claims_premium["Risk_Category"]
    .value_counts()
)

risk_summary
```

```python
loss_ratio_summary = claims_premium[
    [
        "Year",
        "Ultimate_Claims",
        "Gross_Earned_Premium",
        "Loss_Ratio",
        "Risk_Category"
    ]
].copy()

loss_ratio_summary
```

## Summary

The loss ratio was calculated by comparing ultimate claim costs with
gross earned premium for each year.

A higher loss ratio indicates that ultimate claims represent a larger
proportion of earned premium, while a lower loss ratio indicates a
smaller proportion.

The analysis identifies the years with the highest and lowest loss
ratios and classifies each year using analytical risk thresholds.

These results help identify periods of relatively higher or lower
claims pressure compared with premium.
